```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix
from sklearn.datasets import load_wine

_raw = load_wine()
_col_map = {
    'alcohol': 'Alcohol', 'malic_acid': 'Malicacid', 'ash': 'Ash',
    'alcalinity_of_ash': 'Alcalinity_of_ash', 'magnesium': 'Magnesium',
    'total_phenols': 'Total_phenols', 'flavanoids': 'Flavanoids',
    'nonflavanoid_phenols': 'Nonflavanoid_phenols', 'proanthocyanins': 'Proanthocyanins',
    'color_intensity': 'Color_intensity', 'hue': 'Hue',
    'od280/od315_of_diluted_wines': '0D280_0D315_of_diluted_wines', 'proline': 'Proline'
}

X = pd.DataFrame(_raw.data, columns=[_col_map[c] for c in _raw.feature_names])
y = pd.DataFrame({'class': _raw.target + 1})
y_flat = y.values.ravel()

print("Ukuran X:", X.shape, "| Ukuran y:", y.shape)
```

```python
print("=== PRATINJAU FITUR (X) ===")
display(X.head())

print("\n=== PRATINJAU TARGET (y) ===")
display(y.head())

print("\n=== DISTRIBUSI KELAS (seluruh dataset) ===")
classes_full, counts_full = np.unique(y_flat, return_counts=True)
for c, n in zip(classes_full, counts_full):
    print(f"Kelas {c}: {n} sampel")
```

```python
print("=== INFORMASI DATASET ===")
print(X.info())

print("\n=== JUMLAH NILAI KOSONG (MISSING VALUES) ===")
print(X.isnull().sum())

print("\n=== STATISTIK DESKRIPTIF ===")
display(X.describe())
```

```python
baseline_full = counts_full.max() / len(y_flat)
majority_class_full = classes_full[np.argmax(counts_full)]

print("=== BASELINE ACCURACY (seluruh dataset, hanya referensi) ===")
print(f"Kelas Mayoritas: Kelas {majority_class_full} dengan {counts_full.max()} sampel dari {len(y_flat)} total sampel.")
print(f"Baseline Accuracy (Zero-R) - seluruh dataset: {baseline_full * 100:.2f}%")
```

```python
plt.figure(figsize=(6, 4))
sns.countplot(x='class', data=y, hue='class', palette='viridis', legend=False)
plt.title('Distribusi Kelas Target (Wine Class)')
plt.xlabel('Kelas')
plt.ylabel('Jumlah')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 8))
sns.heatmap(X.corr(), annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Heatmap Korelasi Fitur')
plt.tight_layout()
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y_flat, test_size=0.2, random_state=42, stratify=y_flat
)

classes_test, counts_test = np.unique(y_test, return_counts=True)
baseline_test = counts_test.max() / len(y_test)
majority_class_test = classes_test[np.argmax(counts_test)]

print("=== BASELINE ACCURACY (y_test — INI yang dipakai untuk perbandingan akhir) ===")
print(f"Kelas Mayoritas di test set: Kelas {majority_class_test} dengan {counts_test.max()} dari {len(y_test)} sampel.")
print(f"Baseline Accuracy (Zero-R) pada test set: {baseline_test * 100:.2f}%")

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"\nUkuran X_train: {X_train.shape}")
print(f"Ukuran X_test : {X_test.shape}")
```

```python
import sklearn
print("Versi scikit-learn:", sklearn.__version__)

model = LogisticRegression(max_iter=150, random_state=42)
model.fit(X_train_scaled, y_train)

y_pred = model.predict(X_test_scaled)
accuracy = accuracy_score(y_test, y_pred)
print(f"Akurasi Model (Test Accuracy): {accuracy * 100:.2f}%")

print("\n=== LAPORAN KLASIFIKASI ===")
print(classification_report(y_test, y_pred))
```

```python
cm = confusion_matrix(y_test, y_pred, labels=model.classes_)
plt.figure(figsize=(6, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=model.classes_, yticklabels=model.classes_)
plt.title('Confusion Matrix')
plt.xlabel('Prediksi')
plt.ylabel('Aktual')
plt.tight_layout()
plt.show()
```

```python
train_accuracy = accuracy_score(y_train, model.predict(X_train_scaled))

print("=== EVALUASI PERFORMA & VALIDASI ===")
print(f"Baseline Accuracy (y_test, Zero-R)  : {baseline_test * 100:.2f}%")
print(f"Akurasi Data Latih (Train Accuracy) : {train_accuracy * 100:.2f}%")
print(f"Akurasi Data Uji (Test Accuracy)    : {accuracy * 100:.2f}%")

gap = train_accuracy - accuracy
print(f"Gap Train-Test: {gap * 100:.2f} poin persentase")
print(f"Peningkatan di atas baseline test: {(accuracy - baseline_test) * 100:.2f} poin persentase")

pipeline_cv = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LogisticRegression(max_iter=150, random_state=42))
])
scores = cross_val_score(pipeline_cv, X, y_flat, cv=5)
print(f"\nAkurasi 5-Fold CV (Bebas Leakage): {scores.mean() * 100:.2f}% (+/- {scores.std() * 100:.2f}%)")

print(f"\nJumlah iterasi solver (single-split): {model.n_iter_[0]} (batas max_iter: {model.max_iter})")
if model.n_iter_[0] < model.max_iter:
    print("Status Solver: konvergen sebelum mencapai batas maksimum iterasi.")
else:
    print("Status Solver: PERINGATAN - mungkin belum konvergen, naikkan max_iter.")
```

```python
C_values = [0.01, 0.1, 1.0, 10.0, 100.0]
results = []
for C in C_values:
    m = LogisticRegression(max_iter=500, random_state=42, C=C)
    m.fit(X_train_scaled, y_train)
    tr = accuracy_score(y_train, m.predict(X_train_scaled))
    te = accuracy_score(y_test, m.predict(X_test_scaled))
    results.append({'C': C, 'Train Accuracy': tr, 'Test Accuracy': te})

df_c = pd.DataFrame(results)
display(df_c)

plt.figure(figsize=(7, 4))
plt.plot(df_c['C'], df_c['Train Accuracy'], marker='o', label='Train Accuracy')
plt.plot(df_c['C'], df_c['Test Accuracy'], marker='o', label='Test Accuracy')
plt.xscale('log')
plt.xlabel('C (log scale) — semakin besar C, semakin lemah regularisasi')
plt.ylabel('Akurasi')
plt.title('Efek Capacity Dial (C) terhadap Train vs Test Accuracy')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
coef_df = pd.DataFrame(
    model.coef_,
    columns=X.columns,
    index=[f"Kelas {c} vs lainnya" for c in model.classes_]
)
display(coef_df.round(3))

mean_abs_coef = coef_df.abs().mean(axis=0).sort_values(ascending=False)
print("\n=== 2 fitur dengan pengaruh rata-rata terbesar (|koefisien|, lintas kelas) ===")
print(mean_abs_coef.head(2))
```

```python
proba = model.predict_proba(X_test_scaled)
class3_idx = list(model.classes_).index(3)
proba_class3 = proba[:, class3_idx]

print("=== Default (argmax / setara threshold 0.5 relatif terhadap kelas lain) ===")
pred_default = model.predict(X_test_scaled)
recall_default = ((y_test == 3) & (pred_default == 3)).sum() / (y_test == 3).sum()
print(f"Recall kelas 3 (default): {recall_default:.2f}")

print("\n=== Threshold alternatif: anggap kelas 3 jika P(kelas 3) >= 0.3 ===")
pred_lower = np.where(proba_class3 >= 0.3, 3, pred_default)
recall_lower = ((y_test == 3) & (pred_lower == 3)).sum() / (y_test == 3).sum()
precision_lower = ((pred_lower == 3) & (y_test == 3)).sum() / max((pred_lower == 3).sum(), 1)
print(f"Recall kelas 3 (threshold 0.3)   : {recall_lower:.2f}")
print(f"Precision kelas 3 (threshold 0.3): {precision_lower:.2f}")
```

```python
print("=== KESIMPULAN MULTI-CLASS ===")
print(f"Model Logistic Regression berhasil mencapai akurasi sebesar {accuracy * 100:.2f}% pada test set ({len(y_test)} sampel yang sepenuhnya baru).")
print(f"Baseline Zero-R pada populasi test yang SAMA: {baseline_test * 100:.2f}%.")
print(f"Peningkatan di atas baseline (apple-to-apple, populasi sama): {(accuracy - baseline_test) * 100:.2f} poin persentase.")
print(f"Stabilitas performa dikonfirmasi oleh 5-Fold CV bebas leakage: {scores.mean() * 100:.2f}% (+/- {scores.std() * 100:.2f}%).")
print(f"Gap train-test: {gap * 100:.2f} poin, ini dikategorikan sebagai REASONABLE FIT, bukan overfitting bermasalah, karena dikonfirmasi stabil oleh CV")
```
