```python
print("\n--- 3. GRANULARITAS NUMERIK ---")
num_cols = df.select_dtypes(include=[np.number]).columns
for col in num_cols:
    str_vals = df[col].astype(str)
    decimal_places = str_vals.apply(lambda x: len(x.split('.')[-1]) if '.' in x else 0)
    max_dec = decimal_places.max()

    sorted_uniques = np.sort(df[col].unique())
    if len(sorted_uniques) > 1:
        diffs = np.diff(sorted_uniques)
        min_step = diffs[diffs > 0].min() if len(diffs[diffs > 0]) > 0 else 0
    else:
        min_step = 0

    print(f"  {col:25s}: max_decimal={max_dec}, min_step={min_step:.4f}, unique_count={len(sorted_uniques)}")
```

```python
print("\n--- 4. SPEKTRUM FREKUENSI NILAI (TOP 20 PALING SERING MUNCUL) ---")
for col in ['bmi', 'HbA1c_level', 'blood_glucose_level', 'age']:
    freq_spectrum = df[col].value_counts().head(20)
    print(f"\n[{col}] Top 20 nilai paling sering:")
    for rank, (val, cnt) in enumerate(freq_spectrum.items(), 1):
        pct = cnt / len(df) * 100
        print(f"  #{rank:>2}: {val:>10} → {cnt:>5} kali ({pct:.3f}%)")
```

```python
print("\n--- 5. DETEKSI NILAI TERLALU BULAT (INDIKATOR SINTETIK) ---")
for col in ['bmi', 'HbA1c_level', 'blood_glucose_level', 'age']:
    is_round = (df[col] % 1 == 0) | (df[col] % 0.5 == 0)
    round_pct = is_round.mean() * 100

    one_dec = (df[col] * 10) % 1 == 0
    one_dec_pct = one_dec.mean() * 100

    print(f"  {col:25s}: bulat_penuh={round_pct:.1f}% | ≤1_desimal={one_dec_pct:.1f}%")
```

```python
print("\n--- 6. KO-OKURENSI ANOMALI PER BARIS ---")
df['_flag_hba1c_low'] = (df['HbA1c_level'] < 4.0).astype(int)
df['_flag_glucose_cap'] = (df['blood_glucose_level'] == 300).astype(int)
df['_flag_age_infant'] = (df['age'] < 1).astype(int)
df['_flag_bmi_default'] = (df['bmi'] == 27.32).astype(int)
df['_flag_smoking_noinfo'] = (df['smoking_history'] == 'No Info').astype(int)

flag_cols = [c for c in df.columns if c.startswith('_flag_')]
df['_anomaly_count'] = df[flag_cols].sum(axis=1)

print("\nDistribusi jumlah anomali per baris:")
anom_dist = df['_anomaly_count'].value_counts().sort_index()
for n_flags, cnt in anom_dist.items():
    pct = cnt / len(df) * 100
    print(f"  {int(n_flags)} flags: {cnt:>6} baris ({pct:.2f}%)")

print("\nBaris dengan ≥3 anomali simultan:")
multi_anomaly = df[df['_anomaly_count'] >= 3][flag_cols + ['age','bmi','HbA1c_level','blood_glucose_level','diabetes']]
print(f"  Jumlah: {len(multi_anomaly)} baris")
if len(multi_anomaly) > 0:
    display(multi_anomaly.head(20))
```

```python
print("\n--- 7. ENTROPY DIABETES PER KOMBINASI FITUR KLINIS ---")
from scipy.stats import entropy as sp_entropy

clinical_combos = df.groupby(['HbA1c_level', 'blood_glucose_level'])['diabetes'].agg(['mean', 'count', 'std'])
clinical_combos = clinical_combos[clinical_combos['count'] >= 5]

def binary_entropy(p):
    if p == 0 or p == 1:
        return 0.0
    return -p * np.log2(p) - (1 - p) * np.log2(1 - p)

clinical_combos['entropy'] = clinical_combos['mean'].apply(binary_entropy)

print("\nSubgrup dengan entropy TERTINGGI (paling ambigu/tidak pasti):")
high_entropy = clinical_combos.nlargest(10, 'entropy')
display(high_entropy)

print("\nSubgrup dengan entropy TERENDAH & count≥10 (paling deterministik/bocor):")
low_entropy = clinical_combos[clinical_combos['count'] >= 10].nsmallest(10, 'entropy')
display(low_entropy)
```

```python
print("\n--- 8. POLA BERDASARKAN POSISI INDEX ---")

diabetes_idx = df[df['diabetes'] == 1].index
no_diabetes_idx = df[df['diabetes'] == 0].index

d_idx = diabetes_idx.to_numpy().astype(float)
nd_idx = no_diabetes_idx.to_numpy().astype(float)

print(f"  Index diabetes=1: min={int(d_idx.min())}, max={int(d_idx.max())}, mean={d_idx.mean():.0f}")
print(f"  Index diabetes=0: min={int(nd_idx.min())}, max={int(nd_idx.max())}, mean={nd_idx.mean():.0f}")

rolling_prev = df['diabetes'].rolling(window=1000).mean()
print(f"\n  Rolling prevalence (window=1000):")
print(f"    Min: {rolling_prev.min():.4f} di index ~{rolling_prev.idxmin()}")
print(f"    Max: {rolling_prev.max():.4f} di index ~{rolling_prev.idxmax()}")
print(f"    Std of rolling prev: {rolling_prev.std():.4f}")

plt.figure(figsize=(14, 4))
plt.plot(rolling_prev, linewidth=0.8, color='steelblue')
plt.axhline(y=df['diabetes'].mean(), color='red', linestyle='--', label=f'Global mean={df["diabetes"].mean():.3f}')
plt.title("Rolling Prevalence Diabetes (Window=1000)")
plt.xlabel("Index")
plt.ylabel("Prevalence")
plt.legend()
plt.tight_layout()
plt.show()
```

# 3. Data Cleaning

```python
df_clean = df.copy()
print(f"\nData awal: {df_clean.shape[0]} baris, {df_clean.shape[1]} kolom")
```

```python
before = len(df_clean)
df_clean.drop_duplicates(inplace=True)
after = len(df_clean)
print(f"\n[1] Hapus duplikat: {before} → {after} ({before - after} baris dihapus)")
```

```python
df_clean['bmi_is_imputed'] = (df_clean['bmi'] == 27.32).astype(int)
imputed_count = df_clean['bmi_is_imputed'].sum()
print(f"[2] Flag bmi_is_imputed: {imputed_count} baris ditandai (tidak dihapus)")
```

```python
mask_child_noinfo = (df_clean['age'] < 18) & (df_clean['smoking_history'] == 'No Info')
child_converted = mask_child_noinfo.sum()
df_clean.loc[mask_child_noinfo, 'smoking_history'] = 'never'
```

```python
adult_noinfo = ((df_clean['age'] >= 18) & (df_clean['smoking_history'] == 'No Info')).sum()
print(f"[3] Smoking 'No Info' → 'never' (age<18): {child_converted} baris dikonversi")
print(f"    Smoking 'No Info' tetap (age≥18): {adult_noinfo} baris dipertahankan")
```

```python
cap_before = (df_clean['blood_glucose_level'] == 300).sum()
df_clean['blood_glucose_level'] = df_clean['blood_glucose_level'].clip(upper=250)
cap_after = (df_clean['blood_glucose_level'] == 250).sum()
print(f"[4] Glucose clip ≤250: {cap_before} baris di-cap dari 300 → 250")
```

```python
df_clean['hba1c_below_physiological'] = (df_clean['HbA1c_level'] < 4.0).astype(int)
low_hba1c = df_clean['hba1c_below_physiological'].sum()
print(f"[5] Flag hba1c_below_physiological (<4.0): {low_hba1c} baris ditandai")
```

```python
df_clean['hypertension'] = df_clean['hypertension'].astype('int8')
df_clean['heart_disease'] = df_clean['heart_disease'].astype('int8')
df_clean['diabetes'] = df_clean['diabetes'].astype('int8')
df_clean['gender'] = df_clean['gender'].astype('category')
df_clean['smoking_history'] = df_clean['smoking_history'].astype('category')

mem_before = df.memory_usage(deep=True).sum() / 1024**2
mem_after = df_clean.memory_usage(deep=True).sum() / 1024**2
print(f"[6] Konversi tipe: memori {mem_before:.1f} MB → {mem_after:.1f} MB ({(1-mem_after/mem_before)*100:.1f}% hemat)")
```

```python
df_clean.reset_index(drop=True, inplace=True)
print(f"[7] Index direset")

print("\n" + "=" * 60)
print("RINGKASAN DATA CLEANING")
print("=" * 60)
print(f"  Baris akhir     : {len(df_clean)}")
print(f"  Kolom asli      : 9")
print(f"  Kolom baru      : {len(df_clean.columns)} (+3 flag columns)")
print(f"  Duplikat dihapus: {before - after}")
print(f"  Flag ditambahkan:")
print(f"  - bmi_is_imputed           : {df_clean['bmi_is_imputed'].sum()} baris")
print(f"  - hba1c_below_physiological: {df_clean['hba1c_below_physiological'].sum()} baris")
print(f"  Glucose di-clip   : {cap_before} baris (300 → 250)")
print(f"  Smoking dikonversi: {child_converted} baris (No Info → never, age < 18)")
print(f"\nKolom final: {list(df_clean.columns)}")
print("=" * 60)
```

```python
print("\n--- P4: VALIDASI KEBOCORAN vs KRITERIA DIAGNOSTIK ---")

df['_clinically_diabetic'] = ((df['HbA1c_level'] >= 6.5) |
                               (df['blood_glucose_level'] >= 126)).astype(int)

leak_check = df.groupby(['HbA1c_level', 'blood_glucose_level'])['diabetes'].agg(['mean', 'count'])
pure_groups_idx = leak_check[(leak_check['mean'] == 1.0) & (leak_check['count'] >= 10)].index

print("Pure groups (diabetes=100%, n≥10) vs Kriteria Diagnostik:")
match_count = 0
non_match = []
for hba1c, glucose in pure_groups_idx:
    meets_criteria = (hba1c >= 6.5) or (glucose >= 126)
    if meets_criteria:
        match_count += 1
    else:
        non_match.append((hba1c, glucose))

print(f"  Memenuhi kriteria diagnostik: {match_count}/{len(pure_groups_idx)}")
print(f"  TIDAK memenuhi (potential leakage murni): {len(non_match)}/{len(pure_groups_idx)}")

if non_match:
    print("\n  Kelompok yang diabetes=100% TANPA kriteria diagnostik:")
    for h, g in non_match[:15]:
        cnt = leak_check.loc[(h, g), 'count']
        print(f"    HbA1c={h}, Glucose={g} → n={cnt}")
    print("  → Ini BUKAN kriteria klinis. Ini artefak generator/leakage.")
else:
    print("  Semua pure groups konsisten dengan kriteria diagnostik.")

tp = ((df['_clinically_diabetic'] == 1) & (df['diabetes'] == 1)).sum()
fp = ((df['_clinically_diabetic'] == 1) & (df['diabetes'] == 0)).sum()
fn = ((df['_clinically_diabetic'] == 0) & (df['diabetes'] == 1)).sum()
tn = ((df['_clinically_diabetic'] == 0) & (df['diabetes'] == 0)).sum()

print(f"\n  Confusion Matrix (Kriteria Klinis vs Label Dataset):")
print(f"    TP={tp}, FP={fp}, FN={fn}, TN={tn}")
precision = tp / (tp + fp) if (tp + fp) > 0 else 0
recall = tp / (tp + fn) if (tp + fn) > 0 else 0
print(f"    Precision={precision:.4f}, Recall={recall:.4f}")
print(f"  → Jika precision≈1.0 & recall≈1.0: label DIABUAT dari threshold ini")
print(f"  → Jika precision<1 atau recall<1: ada noise/sumber label lain")

df.drop(columns=['_clinically_diabetic'], inplace=True)
```
