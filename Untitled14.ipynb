```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
sns.set_theme(style="whitegrid")
```

```python
try:
    df = pd.read_csv("diabetes_prediction_dataset.csv")
    print("Dataset berhasil dibaca.")
except FileNotFoundError:
    raise FileNotFoundError("Dataset tidak ditemukan.")
```

```python
print("\n--- STATISTIK DESKRIPTIF (NUMERIK) ---")
display(df.describe())
```

```python
print("\n--- STATISTIK DESKRIPTIF (KATEGORIK) ---")
display(df.describe(include=['object', 'category']))
```

```python
print("\n--- MISSING VALUES ---")
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({'Jumlah': missing, 'Persen': missing_pct})
display(missing_df[missing_df['Jumlah'] > 0])
if missing.sum() == 0:
    print("Tidak ada missing values.")
```

```python
print("\n--- DUPLIKAT ---")
duplikat = df.duplicated().sum()
print(f"Jumlah baris duplikat: {duplikat}")
```

```python
target_col = 'diabetes'
if target_col in df.columns:
    print(f"\n--- DISTRIBUSI TARGET ({target_col}) ---")
    display(df[target_col].value_counts(normalize=True))

    plt.figure(figsize=(6, 4))
    sns.countplot(data=df, x=target_col, palette="viridis")
    plt.title(f"Distribusi {target_col}")
    plt.show()
else:
    print(f"Kolom target '{target_col}' tidak ditemukan.")
```

```python
numerik = df.select_dtypes(include=[np.number]).columns.tolist()
kategorik = df.select_dtypes(include=['object', 'category']).columns.tolist()
```

```python
if target_col in numerik: numerik.remove(target_col)
if target_col in kategorik: kategorik.remove(target_col)
```

```python
if numerik:
    print("\n--- DISTRIBUSI FITUR NUMERIK ---")
    n_cols = 3
    n_rows = int(np.ceil(len(numerik) / n_cols))
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, 5 * n_rows))
    axes = axes.flatten()

    for i, col in enumerate(numerik):
        # Representasi Teks: Statistik per fitur
        stats = df[col].describe()
        print(f"\n[{col}]")
        print(f"  Mean: {stats['mean']:.2f} | Std: {stats['std']:.2f}")
        print(f"  Min: {stats['min']:.2f} | Median: {stats['50%']:.2f} | Max: {stats['max']:.2f}")
        skewness = df[col].skew()
        print(f"  Skewness: {skewness:.2f} ({'Simetris' if abs(skewness) < 0.5 else 'Miring Kanan' if skewness > 0 else 'Miring Kiri'})")

        # Chart
        sns.histplot(data=df, x=col, kde=True, ax=axes[i], color='steelblue')
        axes[i].set_title(f"Distribusi: {col}")

    for j in range(len(numerik), len(axes)):
        axes[j].set_visible(False)

    plt.tight_layout()
    plt.show()
```

```python
if kategorik and target_col in df.columns:
    print("\n--- FITUR KATEGORIK VS TARGET ---")
    for col in kategorik:
        # Representasi Teks: Crosstab persentase
        print(f"\n[{col}] vs [{target_col}] (%):")
        ct = pd.crosstab(df[col], df[target_col], normalize='index').round(4) * 100
        display(ct)

        # Chart
        plt.figure(figsize=(10, 5))
        sns.countplot(data=df, x=col, hue=target_col, palette="muted")
        plt.title(f"{col} berdasarkan {target_col}")
        plt.xticks(rotation=45)
        plt.tight_layout()
        plt.show()
```

```python
if len(numerik) > 1:
    print("\n--- MATRIKS KORELASI ---")
    cols_for_corr = numerik + [target_col] if target_col in df.columns else numerik
    corr_matrix = df[cols_for_corr].corr()

    upper_tri = corr_matrix.where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
    strong_corr = upper_tri.stack().abs().sort_values(ascending=False).head(5)
    print("Top 5 Korelasi Terkuat (antar fitur):")
    for (feat1, feat2), val in strong_corr.items():
        direction = "Positif" if corr_matrix.loc[feat1, feat2] > 0 else "Negatif"
        print(f"  {feat1} <-> {feat2}: {val:.3f} ({direction})")

    plt.figure(figsize=(12, 10))
    sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", center=0,
                square=True, linewidths=0.5)
    plt.title("Heatmap Korelasi")
    plt.tight_layout()
    plt.show()
```

```python
if numerik:
    print("\n--- DETEKSI OUTLIER ---")
    n_cols = 3
    n_rows = int(np.ceil(len(numerik) / n_cols))
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(18, 5 * n_rows))
    axes = axes.flatten()

    for i, col in enumerate(numerik):
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR
        outliers = ((df[col] < lower) | (df[col] > upper)).sum()
        pct_outliers = (outliers / len(df)) * 100
        print(f"[{col}] Outlier: {outliers} baris ({pct_outliers:.2f}%) | Batas: [{lower:.2f}, {upper:.2f}]")

        sns.boxplot(data=df, y=col, ax=axes[i], color='lightcoral')
        axes[i].set_title(f"Boxplot: {col}")

    for j in range(len(numerik), len(axes)):
        axes[j].set_visible(False)

    plt.tight_layout()
    plt.show()
```

```python
print("--- ANOMALI KLINIS ---")
print(f"HbA1c < 4.0: {(df['HbA1c_level'] < 4.0).sum()} baris")
print(f"Glucose == 300 (cap?): {(df['blood_glucose_level'] == 300).sum()} baris")
print(f"Age < 1 tahun: {(df['age'] < 1).sum()} baris")
```

```python
print("\n--- 'No Info' SMOKING BY AGE GROUP ---")
df['age_group'] = pd.cut(df['age'], bins=[0, 1, 18, 40, 60, 100],
                          labels=['Bayi', 'Anak', 'Dewasa Muda', 'Dewasa Tua', 'Lansia'])
ct_age_smoke = pd.crosstab(df[df['smoking_history']=='No Info']['age_group'],
                            df[df['smoking_history']=='No Info']['diabetes'],
                            normalize='index') * 100
display(ct_age_smoke)
```

```python
print(f"\n--- DUPLIKAT DETAIL ---")
dup_rows = df[df.duplicated(keep=False)]
print(f"Baris terlibat duplikat: {len(dup_rows)}")
display(dup_rows.head(25))
```

```python
print("\n--- INTERAKSI BMI x AGE VS DIABETES ---")
df['bmi_cat'] = pd.qcut(df['bmi'], q=4, labels=['Underweight','Normal','Overweight','Obese'])
interaction = df.groupby(['bmi_cat', 'age_group'])['diabetes'].mean().unstack()
display(interaction.round(4))
sns.heatmap(interaction, annot=True, fmt=".2%", cmap="YlOrRd")
plt.title("Prevalensi Diabetes: BMI x Age Group")
plt.show()
```

```python
print("\n--- CEK CONCEPT DRIFT (PARUH PERTAMA vs KEDUA) ---")
mid = len(df) // 2
first_half = df.iloc[:mid].describe()[['age','bmi','HbA1c_level','blood_glucose_level']]
second_half = df.iloc[mid:].describe()[['age','bmi','HbA1c_level','blood_glucose_level']]
drift_compare = pd.DataFrame({'Paruh_Pertama': first_half.loc['mean'],
                               'Paruh_Kedua': second_half.loc['mean']})
drift_compare['Selisih_%'] = ((drift_compare['Paruh_Kedua'] - drift_compare['Paruh_Pertama'])
                                / drift_compare['Paruh_Pertama'] * 100).round(2)
display(drift_compare)
```
