### 2. Unduh Dataset dan Pisahkan Fitur (`X`) serta Target (`y`)

```python
from ucimlrepo import fetch_ucirepo

# Mengambil dataset Wine (ID: 109)
wine = fetch_ucirepo(id=109)

# Menyimpan data dalam bentuk pandas DataFrame
X = wine.data.features
y = wine.data.targets
```

### 3. Tampilkan Informasi Variabel, Pratinjau Data, dan Baseline Accuracy

Sesuai dengan prinsip dasar machine learning (Week 7, Slide *"Always Compute the Baseline First"*), kita wajib menghitung **Baseline Accuracy** (Zero-R / Majority Class Accuracy) sebelum melatih model apa pun sebagai pembanding performa objektif.

```python
import numpy as np

print("=== INFORMASI VARIABEL ===")
display(wine.variables)

print("\n=== PRATINJAU FITUR (X) ===")
display(X.head())

print("\n=== PRATINJAU TARGET (y) ===")
display(y.head())

# Menghitung Baseline
y_flat = y.values.ravel()
classes, counts = np.unique(y_flat, return_counts=True)
majority_class_idx = np.argmax(counts)
baseline_accuracy = counts[majority_class_idx] / len(y_flat)

print("\n=== BASELINE ACCURACY ===")
print(f"Kelas Mayoritas: Kelas {classes[majority_class_idx]} dengan {counts[majority_class_idx]} sampel dari {len(y_flat)} total sampel.")
print(f"Baseline Accuracy (Zero-R): {baseline_accuracy * 100:.2f}%")
```

### 4. Analisis Data Eksploratif (EDA) & Pembersihan Data
Mari kita periksa statistik deskriptif dari dataset, serta memeriksa apakah terdapat nilai kosong (*missing values*) yang perlu dibersihkan.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Cek info dataset dan nilai kosong
print("=== INFORMASI DATASET ===")
print(X.info())

print("\n=== JUMLAH NILAI KOSONG (MISSING VALUES) ===")
print(X.isnull().sum())

# 2. Statistik Deskriptif
print("\n=== STATISTIK DESKRIPTIF ===")
display(X.describe())
```

### 5. Distribusi Target & Korelasi Fitur
Mari kita visualisasikan distribusi kelas target dan korelasi antarfiter untuk memahami hubungan datanya.

```python
# Visualisasi distribusi target
plt.figure(figsize=(6, 4))
sns.countplot(x='class', data=y, palette='viridis')
plt.title('Distribusi Kelas Target (Wine Class)')
plt.xlabel('Kelas')
plt.ylabel('Jumlah')
plt.show()

# Visualisasi korelasi antarfiter
plt.figure(figsize=(10, 8))
sns.heatmap(X.corr(), annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Heatmap Korelasi Fitur')
plt.tight_layout()
plt.show()
```

### 6. Pembagian Data (*Splitting*) & Standardisasi
Karena tidak ada nilai kosong (*missing values*), kita bisa langsung melanjutkan ke proses pembagian data menjadi data latih (*training*) dan data uji (*testing*), lalu menerapkan standardisasi menggunakan `StandardScaler`.

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import numpy as np

# Meratakan target y menjadi array 1D agar kompatibel dengan scikit-learn
y_flat = y.values.ravel()

# Membagi data (80% latih, 20% uji)
X_train, X_test, y_train, y_test = train_test_split(
    X, y_flat, test_size=0.2, random_state=42, stratify=y_flat
)

# Standardisasi fitur
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

print(f"Ukuran X_train: {X_train.shape}")
print(f"Ukuran X_test: {X_test.shape}")
```

### 7. Melatih Model Logistic Regression
Kita akan melatih model Logistic Regression. Kita dapat mengatur jumlah iterasi maksimum (`max_iter`) yang berfungsi mirip dengan jumlah *epoch* untuk memastikan algoritma optimasinya konvergen dengan baik.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, accuracy_score, confusion_matrix

# Inisiasi model dengan batasan iterasi (epoch) tertentu, misalnya max_iter=150
# Kita juga menggunakan multi_class='multinomial' karena dataset ini memiliki 3 kelas target
model = LogisticRegression(max_iter=150, random_state=42)

# Melatih model
model.fit(X_train_scaled, y_train)

# Prediksi pada data uji
y_pred = model.predict(X_test_scaled)

# Evaluasi performa model
accuracy = accuracy_score(y_test, y_pred)
print(f"Akurasi Model: {accuracy * 100:.2f}%")

print("\n=== LAPORAN KLASIFIKASI ===")
print(classification_report(y_test, y_pred))

# Matriks Kebingungan (Confusion Matrix)
plt.figure(figsize=(6, 4))
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap='Blues',
            xticklabels=np.unique(y_flat), yticklabels=np.unique(y_flat))
plt.title('Confusion Matrix')
plt.xlabel('Prediksi')
plt.ylabel('Aktual')
plt.show()
```

### 8. Evaluasi Mendalam, Validasi Bebas Leakage & Interpretasi Konteks (Week 7 & Week 9)

Mari kita evaluasi model dengan membandingkan performanya terhadap baseline, mengecek generalisasi (*train-test gap*), dan menggunakan Cross-Validation dengan `Pipeline` untuk mencegah kebocoran data (*data leakage*).

#### Analisis Konteks Precision & Recall (Week 7, Slide hal. 33):
Berdasarkan laporan klasifikasi pada langkah sebelumnya:
* **Kelas 1 & 2** mencapai precision dan recall yang sangat tinggi (sampai 1.00).
* **Kelas 3** memiliki recall sebesar **0.90** (1 dari 10 sampel kelas 3 diprediksi salah sebagai Kelas 2). Hal ini berarti terdapat risiko kecil model gagal mendeteksi keberadaan kultivar anggur tipe 3 sebesar 10%. Dalam konteks klasifikasi komersial anggur, hal ini menunjukkan bahwa model sangat presisi dalam melabeli tipe premium (tidak ada anggur lain yang salah dilabeli sebagai Kelas 3), namun sedikit konservatif dalam mendeteksi anggur Kelas 3 itu sendiri.

```python
from sklearn.model_selection import cross_val_score
from sklearn.pipeline import Pipeline

print(f"=== EVALUASI PERFORMA & VALIDASI ===")
print(f"Baseline Accuracy (Zero-R): {baseline_accuracy * 100:.2f}%")

# 1. Perbandingan Akurasi Train vs Test (Deteksi Overfitting)
train_accuracy = accuracy_score(y_train, model.predict(X_train_scaled))
print(f"Akurasi Data Latih (Train Accuracy): {train_accuracy * 100:.2f}%")
print(f"Akurasi Data Uji (Test Accuracy)   : {accuracy * 100:.2f}%")

gap = train_accuracy - accuracy
print(f"Gap Train-Test: {gap * 100:.2f}% (Peningkatan sebesar {(accuracy - baseline_accuracy)*100:.2f} poin persentase di atas baseline)")

# 2. 5-Fold Cross-Validation menggunakan Pipeline (Mencegah Data Leakage)
pipeline_cv = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LogisticRegression(max_iter=150, random_state=42))
])

scores = cross_val_score(pipeline_cv, X, y_flat, cv=5)
print(f"Akurasi 5-Fold CV (Bebas Leakage): {scores.mean() * 100:.2f}% (+/- {scores.std() * 100:.2f}%)")

# 3. Verifikasi Konvergensi
print(f"Jumlah iterasi solver (Single-Split): {model.n_iter_[0]} (Batas maksimum max_iter: {model.max_iter})")
if model.n_iter_[0] < model.max_iter:
    print("Status Solver: Konvergen sepenuhnya sebelum mencapai batas maksimum iterasi (pada model split utama).")
else:
    print("Status Solver: Peringatan! Model mungkin tidak konvergen.")

# 4. Kesimpulan Hasil Akhir yang "Jujur" (Week 7, Slide hal. 39)
print("\n=== KESIMPULAN AKHIR MULTI-CLASS ===")
print(f"Model Logistic Regression multinomial berhasil mencapai akurasi sebesar {accuracy * 100:.2f}% "
      f"pada test set ({len(y_test)} sampel yang sepenuhnya baru), yang berarti model ini bekerja "
      f"{(accuracy - baseline_accuracy)*100:.2f} poin persentase di atas baseline kelas mayoritas ({baseline_accuracy * 100:.2f}%). "
      f"Stabilitas performa ini dikonfirmasi secara ketat oleh 5-Fold CV bebas leakage sebesar {scores.mean() * 100:.2f}% (+/- {scores.std() * 100:.2f}%).")
```
